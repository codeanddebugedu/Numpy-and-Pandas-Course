```python
import numpy as np
import pandas as pd
```

```python
salary_values = np.array([25000, 32000, 0, 41000, 38000, 29000, 0, 55000, 47000, 36000])
salary_series = pd.Series(salary_values, name="Salary")
salary_series
```

Indexing

```python
salary_series[0]
```

```python
salary_series[4]
```

Slicing

```python
salary_series[0:4]
```

```python
salary_series[:3]
```

```python
salary_series[0:6:2]
```

Printing Multiple Values using Indexing

```python
salary_series[[4,1]]
```

```python
salary_series[[2,5]]
```
